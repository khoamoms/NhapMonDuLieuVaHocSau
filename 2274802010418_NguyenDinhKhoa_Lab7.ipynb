```python
import random
reviews = nltk.corpus.movie_reviews
documents = [(list(reviews.words(fileid)), category)
for category in reviews.categories()
for fileid in reviews.fileids(category)]
random.shuffle(documents)
first_review = ' '.join(documents[0][0])
print(first_review)
```

```python
documents[0][1]

```

```python
all_words = nltk.FreqDist(w.lower() for w in reviews.words())
word_features = list(all_words)
```

```python
def document_features(document, word_features):
 document_words = set(document)
 features = {}
 for word in word_features:features['{}'.format(word)] = (word in document_words)
 return features
```

```python
featuresets = [(document_features(d,word_features), c) for (d,c) in documents]
len(featuresets)
```

```python
train_set, test_set = featuresets[1500:], featuresets[:500]
classifier = nltk.NaiveBayesClassifier.train(train_set)
```

```python
train_set, est_set = featuresets[1500:], featuresets[:500]
classifier = nltk.NaiveBayesClassifier.train(train_set)
print(nltk.classify.accuracy(classifier, test_set))
```

```python
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

van_ban = "This is a very simple sentence showing how to filter out stopwords using NLTK."

# Tách từ và lấy danh sách stopword tiếng Anh
tokens = word_tokenize(van_ban)
sw = set(stopwords.words('english'))

# Lọc bỏ stopword
van_ban_sach = [w for w in tokens if w.lower() not in sw]

print("Văn bản ban đầu:", tokens)
print("Văn bản sau khi lọc:", van_ban_sach)
```

```python
from nltk.corpus import stopwords

# Lấy danh sách stopword mặc định
sw = set(stopwords.words('english'))

# Danh sách các từ bạn MUỐN GIỮ LẠI (bỏ qua không coi là stopword)
tu_muon_giu = {'not', 'no', 'but'}

# Loại bỏ các từ muốn giữ ra khỏi danh sách stopword
sw_tuy_bien = sw - tu_muon_giu

print(f"Số lượng stopword ban đầu: {len(sw)}")
print(f"Số lượng stopword sau khi tinh chỉnh: {len(sw_tuy_bien)}")
print("Từ 'not' có còn là stopword không?", 'not' in sw_tuy_bien)
```

```python
from nltk.corpus import wordnet

tu_can_tim = "computer"
synsets = wordnet.synsets(tu_can_tim)

if synsets:
    dau_tien = synsets[0]
    print(f"Từ: {tu_can_tim}")
    print(f"Định nghĩa (Definition): {dau_tien.definition()}")
    print(f"Ví dụ (Examples): {dau_tien.examples()}")
else:
    print("Không tìm thấy từ này trong WordNet.")
```

```python
from nltk.corpus import wordnet

tu_nhap = "good"
dong_nghia = set()
trai_nghia = set()

for syn in wordnet.synsets(tu_nhap):
    for lemma in syn.lemmas():
        dong_nghia.add(lemma.name()) # Thêm từ đồng nghĩa
        if lemma.antonyms():
            for ant in lemma.antonyms():
                trai_nghia.add(ant.name()) # Thêm từ trái nghĩa

print(f"Từ đồng nghĩa của '{tu_nhap}':", list(dong_nghia)[:10])
print(f"Từ trái nghĩa của '{tu_nhap}':", list(trai_nghia)[:10])
```

```python
from nltk.corpus import wordnet

# Câu 9: So sánh hai danh từ
danh_tu_1 = wordnet.synset('dog.n.01')
danh_tu_2 = wordnet.synset('cat.n.01')
do_tuong_dong_n = danh_tu_1.path_similarity(danh_tu_2)
print(f"Độ tương đồng giữa Danh từ 'dog' và 'cat': {do_tuong_dong_n}")

# Câu 10: So sánh hai động từ
dong_tu_1 = wordnet.synset('run.v.01')
dong_tu_2 = wordnet.synset('walk.v.01')
do_tuong_dong_v = dong_tu_1.path_similarity(dong_tu_2)
print(f"Độ tương đồng giữa Động từ 'run' và 'walk': {do_tuong_dong_v}")
```

```python
import random
from nltk.corpus import names

# Gắn nhãn cho từng tên
danh_sach_nhan = ([(name, 'male') for name in names.words('male.txt')] +
                  [(name, 'female') for name in names.words('female.txt')])

# Trộn ngẫu nhiên
random.seed(42) # Giữ kết quả cố định mỗi lần chạy
random.shuffle(danh_sach_nhan)

print("15 cặp kết hợp ngẫu nhiên đầu tiên (Tên, Nhãn):")
for item in danh_sach_nhan[:15]:
    print(item)
```

```python
from nltk.corpus import names

# Tạo mảng mới chứa: (Kí tự cuối của tên, Nhãn giới tính)
mang_ki_tu_cuoi = ([(name[-1].lower(), 'male') for name in names.words('male.txt')] +
                   [(name[-1].lower(), 'female') for name in names.words('female.txt')])

print("Số lượng phần tử trong mảng mới:", len(mang_ki_tu_cuoi))
print("15 phần tử đầu tiên của mảng mới (Kí tự cuối, Nhãn):")
for item in mang_ki_tu_cuoi[:15]:
    print(item)
```
