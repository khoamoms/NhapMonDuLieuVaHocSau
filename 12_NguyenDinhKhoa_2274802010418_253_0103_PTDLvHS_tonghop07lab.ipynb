LAB 01: BÀI THỰC HÀNH THAO TÁC DỮ LIỆU

```python
import pandas as pd
import numpy as np
import os
import urllib.request

# =====================================================================
# ### 01. Xác định và phân loại dữ liệu định tính và định lượng
# =====================================================================
print("--- ### 01. PHÂN LOẠI DỮ LIỆU ĐỊNH TÍNH & ĐỊNH LƯỢNG ---")
print("-> Định tính: GT, DT, KV, KT")
print("-> Định lượng: T1->N1, T2->N2, T6->N6, DH1, DH2, DH3")

# =====================================================================
# ### 02. Định nghĩa các thang đo phù hợp cho từng biến số
# =====================================================================
print("\n--- ### 02. ĐỊNH NGHĨA CÁC THANG ĐO ---")
print("-> Thang đo Nominal: GT, DT, KT")
print("-> Thang đo Ordinal: KV")
print("-> Thang đo Ratio: Toàn bộ các cột điểm số")

# =====================================================================
# ### 03. Tải dữ liệu lên chương trình và in 10 dòng đầu/cuối
# =====================================================================
print("\n--- ### 03. TẢI DỮ LIỆU VÀ HIỂN THỊ ---")
if not os.path.exists("dulieuxettuyendaihoc.csv"):
    url = "https://raw.githubusercontent.com/datascience-vlu/dataset/main/dulieuxettuyendaihoc.csv"
    urllib.request.urlretrieve(url, "dulieuxettuyendaihoc.csv")
df = pd.read_csv("dulieuxettuyendaihoc.csv")
print("\n10 dòng đầu tiên của dữ liệu gốc:")
display(df.head(10))
print("\n10 dòng cuối cùng của dữ liệu gốc:")
display(df.tail(10))

# =====================================================================
# ### 04. Thống kê và xử lý dữ liệu thiếu cột dân tộc (DT) -> điền 0
# =====================================================================
print("\n--- ### 04. XỬ LÝ DỮ LIỆU THIẾU CỘT DÂN TỘC (DT) ---")
df['DT'] = df['DT'].fillna('0')
print("Các giá trị riêng biệt của DT sau khi xử lý:", df['DT'].unique())

# =====================================================================
# ### 05. Thống kê và xử lý dữ liệu thiếu cột T1 -> dùng Mean
# =====================================================================
print("\n--- ### 05. XỬ LÝ DỮ LIỆU THIẾU CỘT T1 ---")
t1_mean = df['T1'].mean()
df['T1'] = df['T1'].fillna(t1_mean)
print(f"Điểm trung bình cột T1 dùng để thay thế: {t1_mean:.4f}")

# =====================================================================
# ### 06. Xử lý tự động dữ liệu thiếu cho các biến điểm số còn lại
# =====================================================================
print("\n--- ### 06. XỬ LÝ TỰ ĐỘNG CÁC CỘT ĐIỂM KHUYẾT THIẾU CÒN LẠI ---")
diem_cols = ['T2', 'L2', 'H2', 'S2', 'V2', 'X2', 'D2', 'N2', 'T6', 'L6', 'H6', 'S6', 'V6', 'X6', 'D6', 'N6', 'DH1', 'DH2', 'DH3']
for col in diem_cols:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].mean())
print("-> Đã điền xong dữ liệu thiếu cho tất cả các biến điểm số còn lại bằng phương pháp Mean!")

# =====================================================================
# ### 07. Tạo các biến trung bình môn TBM1, TBM2, TBM3
# =====================================================================
print("\n--- ### 07. TÍNH ĐIỂM TRUNG BÌNH MÔN CÁC NĂM (TBM1, TBM2, TBM3) ---")
df['TBM1'] = (df['T1']*2 + df['L1'] + df['H1'] + df['S1'] + df['V1']*2 + df['X1'] + df['D1'] + df['N1']) / 10
df['TBM2'] = (df['T2']*2 + df['L2'] + df['H2'] + df['S2'] + df['V2']*2 + df['X2'] + df['D2'] + df['N2']) / 10
df['TBM3'] = (df['T6']*2 + df['L6'] + df['H6'] + df['S6'] + df['V6']*2 + df['X6'] + df['D6'] + df['N6']) / 10
display(df[['TBM1', 'TBM2', 'TBM3']].head())

# =====================================================================
# ### 08. Tạo biến xếp loại XL1, XL2, XL3
# =====================================================================
print("\n--- ### 08. XẾP LOẠI HỌC LỰC (XL1, XL2, XL3) ---")
def phan_loai(tbm):
    if tbm < 5.0: return 'Y'
    elif tbm < 6.5: return 'TB'
    elif tbm < 8.0: return 'K'
    elif tbm < 9.0: return 'G'
    else: return 'XS'
df['XL1'] = df['TBM1'].apply(phan_loai)
df['XL2'] = df['TBM2'].apply(phan_loai)
df['XL3'] = df['TBM3'].apply(phan_loai)
display(df[['TBM1', 'XL1', 'TBM2', 'XL2', 'TBM3', 'XL3']].head())

# =====================================================================
# ### 09. Chuyển đổi sang điểm hệ Mỹ thang 4 (US_TBM1, 2, 3)
# =====================================================================
print("\n--- ### 09. CHUYỂN ĐỔI SANG ĐIỂM HỆ MỸ THANG 4 (US_TBM) ---")
df['US_TBM1'] = (df['TBM1'] / 10.0) * 4.0
df['US_TBM2'] = (df['TBM2'] / 10.0) * 4.0
df['US_TBM3'] = (df['TBM3'] / 10.0) * 4.0
display(df[['TBM1', 'US_TBM1', 'TBM2', 'US_TBM2', 'TBM3', 'US_TBM3']].head())

# =====================================================================
# ### 10. Tạo biến kết quả xét tuyển (KQXT)
# =====================================================================
print("\n--- ### 10. TÍNH TOÁN KẾT QUẢ XÉT TUYỂN (KQXT) ---")
def tinh_kqxt(row):
    kt = row['KT']
    dh1, dh2, dh3 = row['DH1'], row['DH2'], row['DH3']
    if kt in ['A', 'A1']: diem_xt = (dh1*2 + dh2 + dh3) / 4
    elif kt == 'B': diem_xt = (dh1 + dh2*2 + dh3) / 4
    else: diem_xt = (dh1 + dh2 + dh3) / 3
    return '1' if diem_xt >= 5.0 else '0'
df['KQXT'] = df.apply(tinh_kqxt, axis=1)
display(df[['KT', 'DH1', 'DH2', 'DH3', 'KQXT']].head(10))

# =====================================================================
# ### 11. Lưu trữ dữ liệu ra file processed_dulieuxettuyendaihoc.csv
# =====================================================================
print("\n--- ### 11. XUẤT KẾT QUẢ LƯU TRỮ ---")
df.to_csv("processed_dulieuxettuyendaihoc.csv", index=False)
print("👉 Đã hoàn thành Lab 01 thành công!")
```

LAB 02: BÀI THỰC HÀNH TRÌNH BÀY DỮ LIỆU

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Đọc file dữ liệu đã được xử lý sạch và chuẩn hóa ở Lab 01
# (Đảm bảo file processed_dulieuxettuyendaihoc.csv đã có sẵn trong thư mục làm việc)
df_lab2 = pd.read_csv("processed_dulieuxettuyendaihoc.csv")


# =====================================================================
# ## PHẦN 01: THỐNG KÊ DỮ LIỆU
# =====================================================================

# ### 01. Hãy sắp xếp dữ liệu điểm DH1 theo thứ tự tăng dần
print("--- ### 01. SẮP XẾP ĐIỂM DH1 TĂNG DẦN ---")
df_sorted_dh1 = df_lab2.sort_values(by='DH1')
display(df_sorted_dh1[['GT', 'KV', 'KT', 'DH1']].head(10))


# ### 02. Hãy sắp xếp dữ liệu điểm DH2 tăng dần theo nhóm giới tính
print("\n--- ### 02. SẮP XẾP ĐIỂM DH2 TĂNG DẦN THEO GIỚI TÍNH ---")
df_sorted_dh2 = df_lab2.sort_values(by=['GT', 'DH2'])
display(df_sorted_dh2[['GT', 'DH2']].head(10))


# Định nghĩa các hàm phụ trợ phục vụ tính toán các khoảng tứ phân vị (Q1, Q2, Q3)
def q1(x): return x.quantile(0.25)
def q2(x): return x.quantile(0.50)  # Median
def q3(x): return x.quantile(0.75)


# ### 03. Hãy tạo pivot-table để thống kê các giá trị count, sum, mean, median, min, max, std, Q1, Q2 và Q3 của DH1 theo KT
print("\n--- ### 03. BẢNG PIVOT THỐNG KÊ DH1 THEO KHỐI THI (KT) ---")
pivot_kt = df_lab2.pivot_table(
    values='DH1',
    index='KT',
    aggfunc=['count', 'sum', 'mean', 'median', 'min', 'max', 'std', q1, q2, q3]
)
display(pivot_kt)


# ### 04. Hãy tạo pivot-table để thống kê các giá trị count, sum, mean, median, min, max, std, Q1, Q2 và Q3 của DH1 theo KT và KV
print("\n--- ### 04. BẢNG PIVOT THỐNG KÊ DH1 THEO KT VÀ KHU VỰC (KV) ---")
pivot_kt_kv = df_lab2.pivot_table(
    values='DH1',
    index=['KT', 'KV'],
    aggfunc=['count', 'sum', 'mean', 'median', 'min', 'max', 'std', q1, q2, q3]
)
display(pivot_kt_kv)


# ### 05. Hãy tạo pivot-table để thống kê các giá trị count, sum, mean, median, min, max, std, Q1, Q2 và Q3 của DH1 theo KT, KV và DT
print("\n--- ### 05. BẢNG PIVOT THỐNG KÊ DH1 THEO KT, KV VÀ DÂN TỘC (DT) ---")
pivot_kt_kv_dt = df_lab2.pivot_table(
    values='DH1',
    index=['KT', 'KV', 'DT'],
    aggfunc=['count', 'sum', 'mean', 'median', 'min', 'max', 'std', q1, q2, q3]
)
display(pivot_kt_kv_dt)


# =====================================================================
# ## PHẦN 02: TRÌNH BÀY DỮ LIỆU
# =====================================================================

# ### 01. Hãy trình bày dữ liệu biến: GT (Bảng tần số, tần suất và trực quan hóa)
print("\n--- ### 01. TRÌNH BÀY DỮ LIỆU BIẾN GT ---")
tanso_gt = df_lab2['GT'].value_counts()
tansuat_gt = df_lab2['GT'].value_counts(normalize=True) * 100
thongke_gt = pd.DataFrame({'Tần số': tanso_gt, 'Tần suất (%)': tansuat_gt})
display(thongke_gt)

# Trực quan hóa
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.barplot(x=tanso_gt.index, y=tanso_gt.values, ax=axes[0], palette='pastel', edgecolor='black')
axes[0].set_title('Biểu đồ cột tần số GT')
axes[0].set_ylabel('Số lượng')

axes[1].pie(tansuat_gt, labels=tansuat_gt.index, autopct='%1.1f%%', colors=['lightpink', 'lightblue'])
axes[1].set_title('Biểu đồ tròn tần suất GT')
plt.show()


# ### 02. Hãy trình bày dữ liệu lần lượt các biến: US_TBM1, US_TBM2 và US_TBM3
print("\n--- ### 02. TRÌNH BÀY DỮ LIỆU ĐIỂM US_TBM CÁC NĂM ---")
plt.figure(figsize=(10, 5))
sns.boxplot(data=df_lab2[['US_TBM1', 'US_TBM2', 'US_TBM3']], palette='Set3')
plt.title('Biểu đồ Boxplot so sánh phân phối điểm hệ Mỹ US_TBM')
plt.ylabel('Thang điểm 4')
plt.show()


# ### 03. Hãy trình bày dữ liệu biến DT với các học sinh là nam
print("\n--- ### 03. TRÌNH BÀY BIẾN DT VỚI HỌC SINH NAM ---")
df_nam = df_lab2[df_lab2['GT'] == 'M']
tanso_dt_nam = df_nam['DT'].value_counts()
display(pd.DataFrame({'Số lượng Nam': tanso_dt_nam}))

plt.figure(figsize=(6, 4))
sns.barplot(x=tanso_dt_nam.index, y=tanso_dt_nam.values, palette='Blues_d', edgecolor='black')
plt.title('Cơ cấu Dân tộc của các thí sinh Nam')
plt.ylabel('Số lượng')
plt.show()


# ### 04. Hãy trình bày dữ liệu biến KV với các học sinh là nam thuộc dân tộc Kinh, có điểm thỏa điều kiện
print("\n--- ### 04. TRÌNH BÀY BIẾN KV THEO ĐIỀU KIỆN ĐIỂM SỐ ---")
# Điều kiện: Nam, dân tộc Kinh (ký hiệu '0'), DH1 >= 5.0, DH2 >= 4.0, DH3 >= 4.0
cond_kv = (df_lab2['GT'] == 'M') & (df_lab2['DT'] == '0') & (df_lab2['DH1'] >= 5.0) & (df_lab2['DH2'] >= 4.0) & (df_lab2['DH3'] >= 4.0)
df_kv_filtered = df_lab2[cond_kv]
tanso_kv_cond = df_kv_filtered['KV'].value_counts()
display(pd.DataFrame({'Tần số KV thỏa mãn': tanso_kv_cond}))

if len(df_kv_filtered) > 0:
    plt.figure(figsize=(6, 4))
    sns.barplot(x=tanso_kv_cond.index, y=tanso_kv_cond.values, palette='viridis', edgecolor='black')
    plt.title('Phân bố khu vực thi của thí sinh Nam Kinh đạt điểm chuẩn')
    plt.ylabel('Số lượng')
    plt.show()


# ### 05. Hãy trình bày dữ liệu lần lượt các biến DH1, DH2, DH3 lớn hơn bằng 5.0 và thuộc khu vực 2NT
print("\n--- ### 05. TRÌNH BÀY ĐIỂM SỐ >= 5.0 CỦA THÍ SINH KHU VỰC 2NT ---")
df_2nt = df_lab2[df_lab2['KV'] == '2NT']
dh1_pass = df_2nt[df_2nt['DH1'] >= 5.0]
dh2_pass = df_2nt[df_2nt['DH2'] >= 5.0]
dh3_pass = df_2nt[df_2nt['DH3'] >= 5.0]

df_pass_stats = pd.DataFrame({
    'Môn thi': ['DH1', 'DH2', 'DH3'],
    'Số lượng đạt >= 5.0': [len(dh1_pass), len(dh2_pass), len(dh3_pass)]
})
display(df_pass_stats)


# =====================================================================
# ## PHẦN 03: TRỰC QUAN HÓA DỮ LIỆU THEO NHÓM PHÂN LOẠI
# =====================================================================

# ### 01. Trực quan dữ liệu học sinh nữ trên các nhóm XL1, XL2, XL3 dạng unstacked
print("\n--- ### 01. TRỰC QUAN XẾP LOẠI HỌC SINH NỮ (UNSTACKED) ---")
df_female = df_lab2[df_lab2['GT'] == 'F']
xl_thongke = df_female[['XL1', 'XL2', 'XL3']].apply(pd.Series.value_counts).fillna(0)

xl_thongke.plot(kind='bar', figsize=(10, 5), edgecolor='black')
plt.title('Phân loại học lực của học sinh Nữ qua các năm')
plt.ylabel('Số lượng học sinh')
plt.legend(['Lớp 10 (XL1)', 'Lớp 11 (XL2)', 'Lớp 12 (XL3)'])
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()


# ### 02. Trực quan dữ liệu KQXT trên nhóm học sinh có khối thi A, A1, B thuộc khu vực 1, 2
print("\n--- ### 02. TRỰC QUAN KQXT THEO KHỐI THI VÀ KHU VỰC 1, 2 ---")
cond_p3 = df_lab2['KT'].isin(['A', 'A1', 'B']) & df_lab2['KV'].isin(['1', '2'])
df_p3_filtered = df_lab2[cond_p3]

plt.figure(figsize=(8, 5))
sns.countplot(data=df_p3_filtered, x='KT', hue='KQXT', palette='Set1', edgecolor='black')
plt.title('Tần số Đậu/Rớt theo khối thi (Khu vực 1 & 2)')
plt.xlabel('Khối thi')
plt.legend(['Rớt (0)', 'Đậu (1)'])
plt.show()


# ### 03. Trực quan dữ liệu số lượng thí sinh từng khu vực dựa trên từng nhóm khối thi
print("\n--- ### 03. TRỰC QUAN SỐ LƯỢNG THÍ SINH KV THEO KHỐI THI ---")
plt.figure(figsize=(10, 5))
sns.countplot(data=df_lab2, x='KT', hue='KV', palette='pastel', edgecolor='black')
plt.title('Số thí sinh của từng khu vực trong nhóm khối thi')
plt.xlabel('Khối thi')
plt.legend(title='Khu vực')
plt.show()


# ### 04. Trực quan dữ liệu số lượng thí sinh đậu, rớt trên từng nhóm khối thi
print("\n--- ### 04. TRỰC QUAN KQXT TRÊN TỪNG NHÓM KHỐI THI ---")
plt.figure(figsize=(8, 5))
sns.countplot(data=df_lab2, x='KT', hue='KQXT', palette='RdYlGn', edgecolor='black')
plt.title('Tỉ lệ Đậu/Rớt trên từng nhóm khối thi')
plt.xlabel('Khối thi')
plt.legend(['Rớt (0)', 'Đậu (1)'])
plt.show()


# ### 05. Trực quan dữ liệu số lượng thí sinh đậu rớt trên từng nhóm khu vực
print("\n--- ### 05. TRỰC QUAN KQXT TRÊN TỪNG NHÓM KHU VỰC ---")
plt.figure(figsize=(8, 5))
sns.countplot(data=df_lab2, x='KV', hue='KQXT', palette='coolwarm', edgecolor='black')
plt.title('Tỉ lệ Đậu/Rớt trên từng nhóm khu vực thi')
plt.xlabel('Khu vực')
plt.legend(['Rớt (0)', 'Đậu (1)'])
plt.show()


# ### 06. Trực quan dữ liệu số lượng thí sinh đậu rớt dựa trên từng nhóm dân tộc
print("\n--- ### 06. TRỰC QUAN KQXT TRÊN TỪNG NHÓM DÂN TỘC ---")
plt.figure(figsize=(8, 5))
sns.countplot(data=df_lab2, x='DT', hue='KQXT', palette='Set2', edgecolor='black')
plt.title('Tỉ lệ Đậu/Rớt trên từng nhóm dân tộc')
plt.xlabel('Dân tộc')
plt.legend(['Rớt (0)', 'Đậu (1)'])
plt.show()


# ### 07. Trực quan dữ liệu số lượng thí sinh đậu rớt dựa trên từng nhóm giới tính
print("\n--- ### 07. TRỰC QUAN KQXT TRÊN TỪNG NHÓM GIỚI TÍNH ---")
plt.figure(figsize=(8, 5))
sns.countplot(data=df_lab2, x='GT', hue='KQXT', palette='vlag', edgecolor='black')
plt.title('Tỉ lệ Đậu/Rớt trên từng giới tính')
plt.xlabel('Giới tính')
plt.legend(['Rớt (0)', 'Đậu (1)'])
plt.show()


# =====================================================================
# ## PHẦN 04: TRỰC QUAN HÓA DỮ LIỆU NÂN CAO
# =====================================================================

# ### 01. Vẽ biểu đồ đường Simple cho biến T1
print("\n--- ### 01. BIỂU ĐỒ ĐƯỜNG SIMPLE LINE CHO BIẾN T1 ---")
t1_counts = df_lab2['T1'].value_counts().sort_index()

plt.figure(figsize=(12, 5))
plt.plot(t1_counts.index, t1_counts.values, marker='o', color='b', linestyle='-')
plt.title('Biểu đồ đường thể hiện tần số điểm Toán lớp 10 (T1)')
plt.xlabel('Điểm T1')
plt.ylabel('Số lượng học sinh')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()


# ### 02. Hãy tạo biến phân loại (phanlopt1) cho môn toán (T1)
# a. Từ 0 đến dưới 5 = kém (ký hiệu 'k'); b. Từ 5 đến dưới 7 = trung bình (ký hiệu 'tb')
# c. Từ 7 đến dưới 8 = khá (ký hiệu 'kh'); d. Từ 8 trở lên = giỏi (ký hiệu 'g')
print("\n--- ### 02. PHÂN LOẠI PHANLOPT1 VÀ THỐNG KÊ ---")
def phan_lop_t1(t1):
    if t1 < 5.0: return 'k'
    elif t1 < 7.0: return 'tb'
    elif t1 < 8.0: return 'kh'
    else: return 'g'

df_lab2['phanlopt1'] = df_lab2['T1'].apply(phan_lop_t1)
display(df_lab2[['T1', 'phanlopt1']].head(10))


# ### 03. Lập bảng tần số cho biến phanloait1 (ở đây là phanlopt1)
print("\n--- ### 03. BẢNG TẦN SỐ CHO BIẾN PHANLOPT1 ---")
tanso_phanlop = df_lab2['phanlopt1'].value_counts()
display(pd.DataFrame({'Tần số': tanso_phanlop}))


# ### 04. Vẽ biểu đồ đường Multiple Line cho biến T1 được phân loại bởi biến phanlopt1
print("\n--- ### 04. BIỂU ĐỒ ĐƯỜNG MULTIPLE LINE ---")
multiple_line_data = df_lab2.groupby(['T1', 'phanlopt1']).size().unstack(fill_value=0)

plt.figure(figsize=(14, 6))
for col in ['k', 'tb', 'kh', 'g']:
    if col in multiple_line_data.columns:
        plt.plot(multiple_line_data.index, multiple_line_data[col], marker='s', label=f"Phân lớp: {col}")
plt.title('Biểu đồ Multiple Line của điểm T1 theo các nhóm phân lớp')
plt.xlabel('Điểm T1')
plt.ylabel('Số lượng học sinh')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.show()


# ### 05. Vẽ biểu đồ Drop-line cho biến T1 được phân loại bởi biến phanlopt1
print("\n--- ### 05. BIỂU ĐỒ DROP-LINE CHO BIẾN T1 ---")
# Tạo biểu đồ Drop-line bằng cách kết hợp scatter plot và dựng các đường thẳng đứng đứng xuống y=0
plt.figure(figsize=(14, 6))
# Lấy bảng tần số tổng của biến T1
t1_all_counts = df_lab2['T1'].value_counts().sort_index()

# Vẽ các đường dọc xuống trục hoành (Drop-line)
plt.vlines(x=t1_all_counts.index, ymin=0, ymax=t1_all_counts.values, colors='gray', linestyle='--', alpha=0.7)

# Vẽ các chấm tròn thể hiện phân lớp bằng màu sắc khác nhau
sns.scatterplot(data=df_lab2.groupby(['T1', 'phanlopt1']).size().reset_index(name='Count'),
                x='T1', y='Count', hue='phanlopt1', s=100, palette='Set1', zorder=5)

plt.title('Biểu đồ Drop-line tần số điểm T1 có phân lớp phanlopt1')
plt.xlabel('Điểm T1')
plt.ylabel('Tần số (Count)')
plt.grid(True, axis='y', linestyle=':', alpha=0.5)
plt.show()

```

LAB 03: LÀM SẠCH DỮ LIỆU CƠ BẢN

```python
import pandas as pd
import numpy as np
import io

# =====================================================================
# ## PHẦN 01: KHỞI TẠO VÀ ĐỌC DỮ LIỆU Y KHOA (PATIENT HEART RATE)
# =====================================================================

# ### 01. Giải quyết vấn đề "Missing header in the csv file"
print("--- ### 01. ĐỌC DỮ LIỆU VÀ THÊM DÒNG TIÊU ĐỀ (HEADER) ---")

# Giả lập nội dung file dữ liệu patient_heart_rate.csv từ tài liệu thực hành
raw_data = """1,Micky Mous,56,70kgs,72,69,71, , ,
2,Scrooge McDuck, ,78kgs,78,79,72, , ,
3,Huey McDuck,19,56kgs, , , ,71,78,75
4,Dewey McDuck,32,78kgs,78,76,75, , ,
5,Scööpy Doò,32,172lbs, , , ,78,76,75
6,Micky Mous,56,70kgs,72,69,71, , ,
7, , , , , , , , ,
"""

# Khai báo danh sách tên cột chuẩn mực theo nghiệp vụ bài toán
column_names = ["Id", "Name", "Age", "Weight", "m0006", "m0612", "m1218", "f0006", "f0612", "f1218"]

# Đọc dữ liệu thô và gán tiêu đề cột
df_lab3 = pd.read_csv(io.StringIO(raw_data), names=column_names, header=None)
print("Dữ liệu thô sau khi được thêm tiêu đề cột:")
display(df_lab3.head(10))


# =====================================================================
# ## PHẦN 02: LÀM SẠCH DỮ LIỆU CƠ BẢN
# =====================================================================

# ### 02. Xử lý cột Name chứa hỗn hợp Firstname và Lastname (Tách làm 2 cột)
print("\n--- ### 02. TÁCH CỘT NAME THÀNH FIRSTNAME VÀ LASTNAME ---")
# Sử dụng hàm str.split tách theo khoảng trắng và drop cột cũ
df_lab3[['Firstname', 'Lastname']] = df_lab3['Name'].str.split(expand=True)
df_lab3.drop('Name', axis=1, inplace=True)
display(df_lab3[['Firstname', 'Lastname']].head())


# ### 03. Chuyển đổi các đơn vị đo lường cột Weight về thành đơn vị chuẩn "kg"
print("\n--- ### 03. CHUẨN HÓA ĐƠN VỊ ĐO LƯỜNG CỘT WEIGHT (LBS SANG KGS) ---")
weight = df_lab3['Weight']
for i in range(0, len(weight)):
    x = str(weight[i]).strip()
    if "lbs" in x:
        x = x.replace("lbs", "")
        float_x = float(x)
        # Công thức quy đổi: 1kg = 2.2 lbs
        y = int(float_x / 2.2)
        weight[i] = str(y) + "kgs"
    elif "kgs" in x:
        weight[i] = x

print("Dữ liệu sau khi đồng nhất đơn vị cân nặng:")
display(df_lab3[['Firstname', 'Lastname', 'Weight']].head())


# ### 04. Vấn đề về xuất hiện dòng dữ liệu rỗng (NaN toàn bộ) -> Xóa bỏ
print("\n--- ### 04. XÓA BỎ CÁC DÒNG DỮ LIỆU RỖNG HOÀN TOÀN ---")
print(f"Số lượng dòng trước khi xóa: {len(df_lab3)}")
df_lab3.dropna(how="all", inplace=True)
print(f"Số lượng dòng sau khi xóa dòng trống: {len(df_lab3)}")


# ### 05. Xử lý các dòng dữ liệu bị trùng lặp thông tin hoàn toàn
print("\n--- ### 05. LOẠI BỎ CÁC DÒNG DỮ LIỆU TRÙNG LẶP (DUPLICATES) ---")
df_lab3.drop_duplicates(subset=['Firstname', 'Lastname', 'Age', 'Weight'], inplace=True)
display(df_lab3)


# ### 06. Xử lý dữ liệu bị ảnh hưởng bởi lỗi kí tự lạ định dạng Non-ASCII
print("\n--- ### 06. LOẠI BỎ CÁC KÍ TỰ LẠ NON-ASCII TRONG TÊN ---")
# Thay thế các ký tự không thuộc bảng mã ASCII chuẩn bằng chuỗi rỗng
df_lab3['Firstname'] = df_lab3['Firstname'].str.replace(r'[^\x00-\x7F]+', '', regex=True)
df_lab3['Lastname'] = df_lab3['Lastname'].str.replace(r'[^\x00-\x7F]+', '', regex=True)
display(df_lab3[['Firstname', 'Lastname']])


# ### 07. Xử lý khuyết thiếu thuộc tính số học (Age) bằng giá trị Mean
print("\n--- ### 07. ĐIỀN KHUYẾT THIẾU CỘT AGE BẰNG PHƯƠNG PHÁP MEAN ---")
# Chuyển đổi cột Age về dạng số trước khi tính toán
df_lab3['Age'] = pd.to_numeric(df_lab3['Age'], errors='coerce')
age_mean = df_lab3['Age'].mean()
df_lab3['Age'] = df_lab3['Age'].fillna(int(age_mean))
print(f"Giá trị tuổi trung bình điền thế: {int(age_mean)}")
display(df_lab3)


# =====================================================================
# ## PHẦN 03: PHÂN RÃ CỘT VÀ BIẾN ĐỔI DỮ LIỆU DẠNG TIDY (MELT)
# =====================================================================

# ### 08. Sử dụng kỹ thuật pd.melt để phân rã cột hỗn hợp [sex_and_time]
print("\n--- ### 08. BIẾN ĐỔI CHUYỂN DỮ LIỆU TỪ DẠNG RỘNG SANG DẠNG DỌC (MELT) ---")
# Melt các cột thời gian và giới tính thành dòng dọc
df_tidy = pd.melt(
    df_lab3,
    id_vars=["Id", "Age", "Weight", "Firstname", "Lastname"],
    value_name="PulseRate",
    var_name="sex_and_time"
).sort_values(['Id', 'Age', 'Weight', 'Firstname', 'Lastname'])

display(df_tidy.head(10))


# ### 09. Tách nội dung chuỗi hỗn hợp ra thành 3 cột độc lập: PulseRate, Sex và Time
print("\n--- ### 09. TÁCH BIẾN GIỚI TÍNH VÀ THỜI GIAN THÀNH CÁC THUỘC TÍNH RIÊNG BIỆT ---")
# Sử dụng Regular Expression trích xuất ký tự chữ (Sex) và cụm số liền sau (khung giờ)
tmp_df = df_tidy["sex_and_time"].str.extract(r"(\D)(\d{2})(\d{2})", expand=True)
tmp_df.columns = ["sex", "hours_lower", "hours_upper"]

# Ghép khung giờ thành dạng nối (hh-hh)
tmp_df["Time"] = tmp_df["hours_lower"] + "-" + tmp_df["hours_upper"]

# Nối bảng kết quả tạm thời vào DataFrame chính
df_tidy = pd.concat([df_tidy, tmp_df], axis=1)

# Loại bỏ các cột trung gian thừa và xử lý ép kiểu số cho nhịp tim
df_tidy.drop(['sex_and_time', 'hours_lower', 'hours_upper'], axis=1, inplace=True)
df_tidy['PulseRate'] = pd.to_numeric(df_tidy['PulseRate'], errors='coerce')

# Xóa các dòng không thu thập được chỉ số nhịp tim (PulseRate bị khuyết thiếu hoàn toàn)
df_tidy.dropna(subset=['PulseRate'], inplace=True)
display(df_tidy.head(10))


# ### 10. Khảo sát tỉ lệ dữ liệu thiếu trên biến huyết áp và điền giá trị bằng phương pháp Mean/Y học ổn định
print("\n--- ### 10. KHẢO SÁT VÀ XỬ LÝ LƯU TRỮ NHỊP TIM/HUYẾT ÁP KHUYẾT THIẾU ---")
# Tính toán chỉ số nhịp tim trung bình theo nhóm giới tính (Sex) để điền bổ sung
df_tidy['PulseRate'] = df_tidy.groupby('sex')['PulseRate'].transform(lambda x: x.fillna(x.mean()))
print("-> Đã kiểm tra tính toàn vẹn và xử lý dữ liệu khuyết thiếu nhịp tim theo phân lớp giới tính thành công.")


# =====================================================================
# ## PHẦN 04: ĐÓNG GÓI VÀ XUẤT FILE SẠCH
# =====================================================================

# ### 11. Rút gọn dữ liệu phù hợp, reindex lại dữ liệu và lưu trữ xuống đĩa thành file patient_heart_rate_clean.csv
print("\n--- ### 11. REINDEX VÀ LƯU FILE KẾT QUẢ SẠCH ---")
df_tidy.reset_index(drop=True, inplace=True)

output_file = "patient_heart_rate_clean.csv"
df_tidy.to_csv(output_file, index=False)
print(f"👉 Quá trình làm sạch dữ liệu hoàn tất! Đã xuất file thành công: '{output_file}'")
display(df_tidy)
```

LAB 4: BÀI THỰC HÀNH CHUẨN BỊ DỮ LIỆU

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# =====================================================================
# ## PHẦN 01: DATA CLEANSING & FEATURE ENGINEERING
# =====================================================================

# ### 01. Viết hàm load_data() để tải dữ liệu lên ứng dụng và hiển thị 10 dòng đầu tiên
print("--- ### 01. TẢI DỮ LIỆU TITANIC VÀ HIỂN THỊ MẪU ---")
def load_data():
    # Sử dụng bộ dữ liệu Titanic chuẩn tích hợp sẵn trong thư viện Seaborn
    df_titanic = sns.load_dataset('titanic')
    # Bổ sung cột tên giả lập dạng 'LastName, Title. FirstName' để phục vụ trọn vẹn bài học tách chuỗi ký tự
    np.random.seed(42)
    prefixes = ['Mr.', 'Mrs.', 'Miss', 'Master.']
    df_titanic['Name'] = [f"PassengerLastName, {np.random.choice(prefixes)} PassengerFirstName_{i}" for i in range(len(df_titanic))]
    return df_titanic

df_lab4 = load_data()
display(df_lab4.head(10))


# ### 02. Thống kê dữ liệu thiếu trên các biến số và trực quan hóa dữ liệu thiếu bằng biểu đồ Heatmap
print("\n--- ### 02. THỐNG KÊ VÀ TRỰC QUAN HÓA DỮ LIỆU KHUYẾT THIẾU ---")
# Thống kê số lượng dòng thiếu bằng bảng số liệu
missing_stats = df_lab4.isnull().sum()
print("Số lượng dòng thiếu chi tiết trên từng cột thuộc tính:")
print(missing_stats[missing_stats > 0])

# Vẽ biểu đồ Heatmap hiển thị vị trí khuyết thiếu dữ liệu (Age, Cabin, Embarked)
plt.figure(figsize=(10, 5))
sns.heatmap(df_lab4.isnull(), cbar=False, cmap='viridis', yticklabels=False)
plt.title('Bản đồ biểu thị các giá trị thiếu (Missing Values) trên Dataset Titanic')
plt.show()

print("> Nhận xét: Thuộc tính 'age' thiếu khoảng 20%, thuộc tính 'deck/cabin' thiếu cực kỳ nghiêm trọng (>77%), 'embarked' chỉ thiếu 2 dòng.")


# ### 03. Xử lý tên cột thuộc tính Name, tách ra làm 2 cột: firstName và secondName và xóa cột Name gốc
print("\n--- ### 03. PHÂN TÁCH CHUỖI CỘT TÊN (NAME) ---")
# Tách LastName và cụm còn lại dựa trên dấu phẩy ','
df_lab4[['firstName', 'secondName']] = df_lab4['Name'].str.split(',', expand=True)
# Xóa cột Name gốc để tối ưu bộ nhớ dữ liệu
df_lab4.drop('Name', axis=1, inplace=True)
display(df_lab4[['firstName', 'secondName']].head())


# ### 04. Xử lý rút gọn kích thước dữ liệu trên cột Sex như sau: thay thế male -> M và female -> F
print("\n--- ### 04. RÚT GỌN ĐỊNH DẠNG THUỘC TÍNH GIỚI TÍNH (SEX) ---")
df_lab4['sex'] = df_lab4['sex'].map({'male': 'M', 'female': 'F'})
display(df_lab4['sex'].value_counts())


# ### 05. Xử lý dữ liệu thiếu trên biến Age bằng cách thay thế bằng giá trị trung vị (Median) theo từng nhóm hạng vé (Pclass)
print("\n--- ### 05. TRỰC QUAN HÓA VÀ XỬ LÝ ĐIỀN KHUYẾT TUỔI THEO PCLASS ---")
# Vẽ biểu đồ Boxplot khảo sát phân phối tuổi theo từng nhóm hạng hành khách (Pclass)
plt.figure(figsize=(8, 5))
sns.boxplot(data=df_lab4, x='pclass', y='age', palette='Set2')
plt.title('Biểu đồ Boxplot khảo sát phân phối tuổi (Age) theo hạng vé (Pclass)')
plt.show()

# Tính toán giá trị trung vị (Median) tuổi của từng hạng vé
median_ages = df_lab4.groupby('pclass')['age'].median()
print("Tuổi trung vị theo từng nhóm hạng vé (Pclass):")
print(median_ages)

# Hàm điền khuyết tuổi dựa theo phân nhóm pclass tương ứng
def impute_age(row):
    if pd.isna(row['age']):
        return median_ages[row['pclass']]
    return row['age']

df_lab4['age'] = df_lab4.apply(impute_age, axis=1)

# Trực quan lại bản đồ khuyết thiếu sau khi xử lý cột Age
plt.figure(figsize=(10, 2))
sns.heatmap(df_lab4[['age']].isnull(), cbar=False, cmap='viridis', yticklabels=False)
plt.title('Kiểm tra khuyết thiếu cột Age sau khi Imputation')
plt.show()


# ### 06. Xây dựng biến số Agegroup có thang đo thứ tự được ánh xạ theo thang đo khoảng dựa trên độ tuổi
print("\n--- ### 06. XÂY DỰNG BIẾN PHÂN LỚP ĐỘ TUỔI (AGEGROUP) ---")
# Phân loại khoảng: <=12 -> Kid, (12-18] -> Teen, (18-60] -> Adult, >60 -> Older
def map_age_group(age):
    if age <= 12: return 'Kid'
    elif age <= 18: return 'Teen'
    elif age <= 60: return 'Adult'
    else: return 'Older'

df_lab4['Agegroup'] = df_lab4['age'].apply(map_age_group)
display(df_lab4['Agegroup'].value_counts())


# ### 07. Tiến hành thêm đặc trưng về danh xưng (namePrefix) bằng cách tách từ secondName
print("\n--- ### 07. TRÚT TRÍCH ĐẶC TRƯNG DANH XƯNG XÃ HỘI (NAMEPREFIX) ---")
# Trích xuất từ nằm trước dấu chấm '.' trong chuỗi secondName
df_lab4['namePrefix'] = df_lab4['secondName'].str.extract(r'([A-Za-z]+)\.', expand=False)
display(df_lab4['namePrefix'].value_counts())


# ### 08. Khai thác thêm thông tin số lượng thành viên đi theo nhóm thân quen (familySize)
print("\n--- ### 08. XÂY DỰNG ĐẶC TRƯNG QUY MÔ GIA ĐÌNH (FAMILYSIZE) ---")
# Quy thức tính toán: familySize = 1 + sibsp + parch
df_lab4['familySize'] = 1 + df_lab4['sibsp'] + df_lab4['parch']
display(df_lab4[['sibsp', 'parch', 'familySize']].head())


# ### 09. Tạo thêm đặc trưng 'Alone' để xác định hành khách đi theo nhóm hay cá nhân dựa trên familySize
print("\n--- ### 09. XÂY DỰNG ĐẶC TRƯNG ĐI MỘT MÌNH (ALONE) ---")
# Nếu familySize == 1 tức là hành khách đi một mình (Alone = 1), ngược lại đi theo nhóm (Alone = 0)
df_lab4['Alone'] = df_lab4['familySize'].apply(lambda x: 1 if x == 1 else 0)
display(df_lab4['Alone'].value_counts())


# ### 10. Tiến hành tách loại cabin (typeCabin) mà hành khách ở, khuyết thiếu điền "Unknown"
print("\n--- ### 10. PHÂN TÁCH VÀ LÀM SẠCH THUỘC TÍNH LOẠI CABIN ---")
# Trong tập dữ liệu của Seaborn, ký hiệu chữ cái đầu tiên nằm ở cột 'deck' hoặc có thể lấy từ chữ đầu tiên của cabin
df_lab4['typeCabin'] = df_lab4['deck'].astype(str).apply(lambda x: 'Unknown' if x == 'nan' else x)
display(df_lab4['typeCabin'].value_counts())


# =====================================================================
# ## PHẦN 02: KHAI THÁC THÔNG TIN HỮU ÍCH – EDA (EXPLORATORY DATA ANALYSIS)
# =====================================================================

# ### 11. Trực quan thông tin tương quan tỉ lệ sống sót (survived) và thiệt mạng trên từng nhóm giới tính
print("\n--- ### 11. EDA: TỈ LỆ SỐNG SÓT THEO GIỚI TÍNH ---")
plt.figure(figsize=(6, 4))
sns.countplot(data=df_lab4, x='sex', hue='survived', palette='Set1', edgecolor='black')
plt.title('Tương quan Tỉ lệ Sống sót (1) / Thiệt mạng (0) theo Giới tính')
plt.xlabel('Giới tính (M: Nam, F: Nữ)')
plt.ylabel('Số lượng hành khách')
plt.legend(['Thiệt mạng (0)', 'Sống sót (1)'])
plt.show()


# ### 12. Trực quan thông tin hành khách sống sót trên từng nhóm phân loại hành khách (Pclass)
print("\n--- ### 12. EDA: TỈ LỆ SỐNG SÓT THEO HẠNG VÉ (PCLASS) ---")
plt.figure(figsize=(6, 4))
sns.countplot(data=df_lab4, x='pclass', hue='survived', palette='viridis', edgecolor='black')
plt.title('Tương quan Tỉ lệ Sống sót theo Hạng hành khách (Pclass)')
plt.xlabel('Hạng vé (Pclass)')
plt.ylabel('Số lượng hành khách')
plt.legend(['Thiệt mạng (0)', 'Sống sót (1)'])
plt.show()


# ### 13. Trực quan thông tin hành khách sống sót trên từng nhóm giới tính và thang đo tuổi tác (Agegroup)
print("\n--- ### 13. EDA: TỈ LỆ SỐNG SÓT THEO GIỚI TÍNH VÀ NHÓM TUỔI ---")
g = sns.catplot(data=df_lab4, x='Agegroup', hue='survived', col='sex', kind='count', palette='coolwarm', edgecolor='black', height=4, aspect=1.2)
g.set_axis_labels("Nhóm tuổi (Agegroup)", "Số lượng hành khách")
g.fig.suptitle('Tỉ lệ Đậu/Rớt hải trình theo Giới tính và Nhóm tuổi', y=1.05, fontweight='bold')
plt.show()


# ### 14. Trực quan xác suất hành khách sống sót dựa trên thông tin nhóm đi cùng (familySize / Alone)
print("\n--- ### 14. EDA: XÁC SUẤT SỐNG SÓT DỰA TRÊN QUY MÔ ĐI CÙNG ---")
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
# Điểm đồ xác suất tuyến tính theo familySize
sns.lineplot(data=df_lab4, x='familySize', y='survived', marker='o', color='purple', ax=axes[0])
axes[0].set_title('Xác suất Sống sót theo Số lượng thành viên đi cùng')
axes[0].set_xlabel('Quy mô gia đình (familySize)')
axes[0].set_ylabel('Tỉ lệ sống sót trung bình')

# Biểu đồ đếm hành khách đi một mình
sns.countplot(data=df_lab4, x='Alone', hue='survived', palette='Set3', edgecolor='black', ax=axes[1])
axes[1].set_title('Tương quan Sống sót giữa nhóm Đi một mình (1) và Đi chung (0)')
axes[1].set_xlabel('Đi một mình (Alone)')
axes[1].set_ylabel('Số lượng hành khách')
axes[1].set_legend(['Thiệt mạng (0)', 'Sống sót (1)'])

plt.tight_layout()
plt.show()


# ### 15. Trực quan số lượng người thiệt mạng và sống sót theo phân lớp (Pclass) hành khách và cảng cập bến
print("\n--- ### 15. EDA: PHÂN TÍCH KẾT HỢP PCLASS VÀ CẢNG CẬP BẾN ---")
# Trong tập dữ liệu thô, cột 'embarked_town' hoặc 'embarked' đại diện cho cảng xuất phát
plt.figure(figsize=(10, 5))
sns.countplot(data=df_lab4, x='embarked', hue='survived', palette='magma', edgecolor='black')
plt.title('Số lượng hành khách Thiệt mạng/Sống sót theo các Cảng xuất phát')
plt.xlabel('Cảng xuất phát (C = Cherbourg, Q = Queenstown, S = Southampton)')
plt.ylabel('Số lượng')
plt.legend(['Thiệt mạng (0)', 'Sống sót (1)'])
plt.show()

print("👉 Hoàn thành toàn bộ Lab 04! Toàn bộ cấu trúc sạch dữ liệu đã sẵn sàng.")
```

Phần 02: Huấn luyện Mạng Nơ-ron Cơ bản với Dữ liệu Gốc

```python
import matplotlib.pyplot as plt
import numpy as np
import torch
import torch.nn as nn
from torch.optim import SGD
from torch.utils.data import DataLoader, Dataset
from torchvision import datasets

device = "cuda" if torch.cuda.is_available() else "cpu"

# Tải dữ liệu
data_folder = "~/data/FMNIST"
fmnist = datasets.FashionMNIST(data_folder, download=True, train=True)
tr_images = fmnist.data
tr_targets = fmnist.targets


# Định nghĩa lớp Dataset
class FMNISTDataset(Dataset):

    def __init__(self, x, y):
        x = x.float()
        x = x.view(-1, 28 * 28)
        self.x, self.y = x, y

    def __getitem__(self, ix):
        x, y = self.x[ix], self.y[ix]
        return x.to(device), y.to(device)

    def __len__(self):
        return len(self.x)


def get_data():
    train = FMNISTDataset(tr_images, tr_targets)
    trn_dl = DataLoader(train, batch_size=32, shuffle=True)
    return trn_dl


def get_model():
    model = nn.Sequential(
        nn.Linear(28 * 28, 1000), nn.ReLU(), nn.Linear(1000, 10)
    ).to(device)
    loss_fn = nn.CrossEntropyLoss()
    optimizer = SGD(model.parameters(), lr=1e-2)
    return model, loss_fn, optimizer


def train_batch(x, y, model, opt, loss_fn):
    model.train()
    prediction = model(x)
    batch_loss = loss_fn(prediction, y)
    batch_loss.backward()
    opt.step()
    opt.zero_grad()
    return batch_loss.item()


@torch.no_grad()
def accuracy(x, y, model):
    model.eval()
    prediction = model(x)
    max_values, argmaxes = prediction.max(-1)
    is_correct = argmaxes == y
    return is_correct.cpu().numpy().tolist()


# Tiến trình huấn luyện
trn_dl = get_data()
model, loss_fn, optimizer = get_model()
losses, accuracies = [], []

for epoch in range(5):
    print(f"Epoch: {epoch}")
    epoch_losses, epoch_accuracies = [], []

    for ix, batch in enumerate(iter(trn_dl)):
        x, y = batch
        batch_loss = train_batch(x, y, model, optimizer, loss_fn)
        epoch_losses.append(batch_loss)

    epoch_loss = np.array(epoch_losses).mean()

    for ix, batch in enumerate(iter(trn_dl)):
        x, y = batch
        is_correct = accuracy(x, y, model)
        epoch_accuracies.extend(is_correct)

    epoch_accuracy = np.mean(epoch_accuracies)
    losses.append(epoch_loss)
    accuracies.append(epoch_accuracy)

# Vẽ biểu đồ kết quả
epochs = np.arange(5) + 1
plt.figure(figsize=(20, 5))
plt.subplot(121)
plt.title("Loss value over increasing epochs")
plt.plot(epochs, losses, label="Training Loss")
plt.legend()

plt.subplot(122)
plt.title("Accuracy value over increasing epochs")
plt.plot(epochs, accuracies, label="Training Accuracy")
plt.gca().set_yticklabels(
    ["{:.0f}%".format(x * 100) for x in plt.gca().get_yticks()]
)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import torch
import torch.nn as nn
from torch.optim import SGD
from torch.utils.data import DataLoader, Dataset
from torchvision import datasets

device = "cuda" if torch.cuda.is_available() else "cpu"

data_folder = "~/data/FMNIST"
fmnist = datasets.FashionMNIST(data_folder, download=True, train=True)
tr_images = fmnist.data
tr_targets = fmnist.targets


# Lớp Dataset có chia tỷ lệ đầu vào (Scaling)
class FMNISTDataset(Dataset):

    def __init__(self, x, y):
        x = x.float() / 255  # Chuẩn hóa về [0, 1]
        x = x.view(-1, 28 * 28)
        self.x, self.y = x, y

    def __getitem__(self, ix):
        x, y = self.x[ix], self.y[ix]
        return x.to(device), y.to(device)

    def __len__(self):
        return len(self.x)


def get_data():
    train = FMNISTDataset(tr_images, tr_targets)
    trn_dl = DataLoader(train, batch_size=32, shuffle=True)
    return trn_dl


def get_model():
    model = nn.Sequential(
        nn.Linear(28 * 28, 1000), nn.ReLU(), nn.Linear(1000, 10)
    ).to(device)
    loss_fn = nn.CrossEntropyLoss()
    optimizer = SGD(model.parameters(), lr=1e-2)
    return model, loss_fn, optimizer


def train_batch(x, y, model, opt, loss_fn):
    model.train()
    prediction = model(x)
    batch_loss = loss_fn(prediction, y)
    batch_loss.backward()
    opt.step()
    opt.zero_grad()
    return batch_loss.item()


def accuracy(x, y, model):
    model.eval()
    with torch.no_grad():
        prediction = model(x)
        max_values, argmaxes = prediction.max(-1)
        is_correct = argmaxes == y
    return is_correct.cpu().numpy().tolist()


# Tiến trình chạy thực nghiệm
trn_dl = get_data()
model, loss_fn, optimizer = get_model()
losses, accuracies = [], []

for epoch in range(5):
    print(f"Epoch: {epoch}")
    epoch_losses, epoch_accuracies = [], []

    for ix, batch in enumerate(iter(trn_dl)):
        x, y = batch
        batch_loss = train_batch(x, y, model, optimizer, loss_fn)
        epoch_losses.append(batch_loss)

    epoch_loss = np.array(epoch_losses).mean()

    for ix, batch in enumerate(iter(trn_dl)):
        x, y = batch
        is_correct = accuracy(x, y, model)
        epoch_accuracies.extend(is_correct)

    epoch_accuracy = np.mean(epoch_accuracies)
    losses.append(epoch_loss)
    accuracies.append(epoch_accuracy)

# Trực quan hóa
epochs = np.arange(5) + 1
plt.figure(figsize=(20, 5))
plt.subplot(121)
plt.title("Loss value over increasing epochs")
plt.plot(epochs, losses, label="Training Loss")
plt.legend()

plt.subplot(122)
plt.title("Accuracy value over increasing epochs")
plt.plot(epochs, accuracies, label="Training Accuracy")
plt.gca().set_yticklabels(
    ["{:.0f}%".format(x * 100) for x in plt.gca().get_yticks()]
)
plt.legend()
plt.show()
```

Phần 04: Ảnh hưởng của Batch Size và Đánh giá trên Tập Validation

```python
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import numpy as np
import torch
import torch.nn as nn
from torch.optim import Adam
from torch.utils.data import DataLoader, Dataset
from torchvision import datasets

device = "cuda" if torch.cuda.is_available() else "cpu"

# Tải cả dữ liệu Train và Validation
data_folder = "~/data/FMNIST"
fmnist = datasets.FashionMNIST(data_folder, download=True, train=True)
tr_images = fmnist.data
tr_targets = fmnist.targets

val_fmnist = datasets.FashionMNIST(data_folder, download=True, train=False)
val_images = val_fmnist.data
val_targets = val_fmnist.targets


class FMNISTDataset(Dataset):

    def __init__(self, x, y):
        x = x.float() / 255
        x = x.view(-1, 28 * 28)
        self.x, self.y = x, y

    def __getitem__(self, ix):
        return self.x[ix].to(device), self.y[ix].to(device)

    def __len__(self):
        return len(self.x)


# Khởi tạo data loader với tham số batch_size tùy chỉnh
def get_data(batch_size=32):
    train = FMNISTDataset(tr_images, tr_targets)
    trn_dl = DataLoader(train, batch_size=batch_size, shuffle=True)
    val = FMNISTDataset(val_images, val_targets)
    val_dl = DataLoader(val, batch_size=len(val_images), shuffle=False)
    return trn_dl, val_dl


def get_model():
    model = nn.Sequential(
        nn.Linear(28 * 28, 1000), nn.ReLU(), nn.Linear(1000, 10)
    ).to(device)
    loss_fn = nn.CrossEntropyLoss()
    optimizer = Adam(model.parameters(), lr=1e-2)
    return model, loss_fn, optimizer


def train_batch(x, y, model, opt, loss_fn):
    model.train()
    prediction = model(x)
    batch_loss = loss_fn(prediction, y)
    batch_loss.backward()
    opt.step()
    opt.zero_grad()
    return batch_loss.item()


def accuracy(x, y, model):
    model.eval()
    with torch.no_grad():
        prediction = model(x)
        max_values, argmaxes = prediction.max(-1)
        is_correct = argmaxes == y
    return is_correct.cpu().numpy().tolist()


@torch.no_grad()
def val_loss(x, y, model):
    prediction = model(x)
    val_loss = loss_fn(prediction, y)
    return val_loss.item()


# Thực hiện huấn luyện (Sửa batch_size=10000 tại đây nếu muốn kiểm tra batch lớn)
trn_dl, val_dl = get_data(batch_size=32)
model, loss_fn, optimizer = get_model()

train_losses, train_accuracies = [], []
val_losses, val_accuracies = [], []

for epoch in range(5):
    print(f"Epoch: {epoch}")
    train_epoch_losses, train_epoch_accuracies = [], []

    for ix, batch in enumerate(iter(trn_dl)):
        x, y = batch
        batch_loss = train_batch(x, y, model, optimizer, loss_fn)
        train_epoch_losses.append(batch_loss)

    train_epoch_loss = np.array(train_epoch_losses).mean()

    for ix, batch in enumerate(iter(trn_dl)):
        x, y = batch
        is_correct = accuracy(x, y, model)
        train_epoch_accuracies.extend(is_correct)
    train_epoch_accuracy = np.mean(train_epoch_accuracies)

    for ix, batch in enumerate(iter(val_dl)):
        x, y = batch
        val_is_correct = accuracy(x, y, model)
        validation_loss = val_loss(x, y, model)
        val_epoch_accuracy = np.mean(val_is_correct)

    train_losses.append(train_epoch_loss)
    train_accuracies.append(train_epoch_accuracy)
    val_losses.append(validation_loss)
    val_accuracies.append(val_epoch_accuracy)

# Vẽ đồ thị so sánh Train vs Validation
epochs = np.arange(5) + 1
plt.subplot(211)
plt.plot(epochs, train_losses, "bo", label="Training loss")
plt.plot(epochs, val_losses, "r", label="Validation loss")
plt.gca().xaxis.set_major_locator(mticker.MultipleLocator(1))
plt.title("Training and validation loss")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.legend()
plt.show()

plt.subplot(212)
plt.plot(epochs, train_accuracies, "bo", label="Training accuracy")
plt.plot(epochs, val_accuracies, "r", label="Validation accuracy")
plt.gca().xaxis.set_major_locator(mticker.MultipleLocator(1))
plt.title("Training and validation accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.gca().set_yticklabels(
    ["{:.0f}%".format(x * 100) for x in plt.gca().get_yticks()]
)
plt.legend()
plt.show()
```

```python
# --- Cấu hình 1: Sử dụng SGD với Learning Rate tùy chọn ---
from torch.optim import SGD


def get_model_sgd(lr_rate=1e-2):
    model = nn.Sequential(
        nn.Linear(28 * 28, 1000), nn.ReLU(), nn.Linear(1000, 10)
    ).to(device)
    loss_fn = nn.CrossEntropyLoss()
    optimizer = SGD(model.parameters(), lr=lr_rate)
    return model, loss_fn, optimizer


# --- Cấu hình 2: Sử dụng Adam với Learning Rate tùy chọn ---
from torch.optim import Adam


def get_model_adam(lr_rate=1e-3):
    model = nn.Sequential(
        nn.Linear(28 * 28, 1000), nn.ReLU(), nn.Linear(1000, 10)
    ).to(device)
    loss_fn = nn.CrossEntropyLoss()
    optimizer = Adam(model.parameters(), lr=lr_rate)
    return model, loss_fn, optimizer
```

LAB 6:XÂY DỰNG MẠNG NƠ-RON SÂU

```python

import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# =====================================================================
# ## PHẦN 01: CHUẨN BỊ DỮ LIỆU
# =====================================================================

# ### 01. Tiền xử lý dữ liệu (Transform) và tải Dataset
print("--- ### 01. TẢI VÀ CHUẨN HÓA DỮ LIỆU ---")
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,)) # Chuẩn hóa ảnh về khoảng [-1, 1]
])

train_dataset = datasets.FashionMNIST(root='./data', train=True, download=True, transform=transform)
test_dataset = datasets.FashionMNIST(root='./data', train=False, download=True, transform=transform)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=64, shuffle=False)

# ### 02. Nhận diện cấu hình GPU/CPU
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Đang chạy mô hình bằng: {device}")

# =====================================================================
# ## PHẦN 02: XÂY DỰNG MÔ HÌNH VÀ HUẤN LUYỆN
# =====================================================================

# ### 03. Định nghĩa cấu trúc mạng nơ-ron FashionDNN
class FashionDNN(nn.Module):
    def __init__(self):
        super(FashionDNN, self).__init__()
        self.fc1 = nn.Linear(28 * 28, 512)
        self.fc2 = nn.Linear(512, 256)
        self.fc3 = nn.Linear(256, 128)
        self.fc4 = nn.Linear(128, 10)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(0.2)

    def forward(self, x):
        x = x.view(-1, 28 * 28) # Phẳng hóa ảnh
        x = self.relu(self.fc1(x))
        x = self.dropout(x)
        x = self.relu(self.fc2(x))
        x = self.dropout(x)
        x = self.relu(self.fc3(x))
        x = self.fc4(x)
        return x

model = FashionDNN().to(device)

# ### 04. Thiết lập hàm mất mát và tối ưu hóa
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

# ### 05. Huấn luyện mô hình (3 Epochs)
print("\n--- ### 05. BẮT ĐẦU HUẤN LUYỆN MÔ HÌNH ---")
epochs = 3
for epoch in range(epochs):
    model.train()
    running_loss = 0.0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * images.size(0)

    epoch_loss = running_loss / len(train_loader.dataset)
    print(f"Vòng lặp (Epoch) {epoch+1}/{epochs} | Loss: {epoch_loss:.4f}")

# =====================================================================
# ## PHẦN 03: ĐÁNH GIÁ MÔ HÌNH
# =====================================================================

# ### 06. Kiểm tra độ chính xác trên tập kiểm tra (Test Set)
model.eval()
correct = 0
total = 0
with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)
        outputs = model(images)
        _, predicted = torch.max(outputs.data, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

print(f"\n--- ### 06. KẾT QUẢ KIỂM TRA ---")
print(f"Độ chính xác đạt được (Accuracy): {100 * correct / total:.2f}%")
```

```python
# =====================================================================
# # LAB 07: PHÂN TÍCH DỮ LIỆU DẠNG VĂN BẢN VỚI NLTK
# =====================================================================
import nltk
import string
import random
from urllib import request
from bs4 import BeautifulSoup
from nltk.corpus import stopwords, names, wordnet
from nltk.tokenize import word_tokenize

# Tải các tài nguyên cần thiết của NLTK
nltk.download('gutenberg')
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('movie_reviews')
nltk.download('names')
nltk.download('wordnet')
nltk.download('tagsets')
nltk.download('punkt_tab')

# =====================================================================
# ## PHẦN 01: GIỚI THIỆU VÀ TÌM TỪ VỚI NLTK
# =====================================================================

# ### 01. Liệt kê tên các file trong kho ngữ liệu Gutenberg
print("--- ### 01. DANH SÁCH FILE GUTENBERG ---")
gb = nltk.corpus.gutenberg
print("Gutenberg files: ", gb.fileids())

# ### 02. Tìm từ với NLTK (concordance)
print("\n--- ### 02. TÌM TỪ VỚI CONCORDANCE (Ví dụ: 'Stage') ---")
macbeth = nltk.corpus.gutenberg.words('shakespeare-macbeth.txt')
text = nltk.Text(macbeth)
text.concordance('Stage')

# =====================================================================
# ## PHẦN 02: PHÂN TÍCH TẦN SỐ VÀ LÀM SẠCH VĂN BẢN
# =====================================================================

# ### 03. Phân tích tần số từ (đã loại bỏ stopword và dấu câu)
print("\n--- ### 03. PHÂN TÍCH TẦN SỐ TỪ ---")
sw = set(stopwords.words('english'))
punctuation = set(string.punctuation)
macbeth_filtered2 = [w.lower() for w in macbeth if w.lower() not in sw and w.lower() not in punctuation]
fd = nltk.FreqDist(macbeth_filtered2)
print("10 từ xuất hiện nhiều nhất (đã làm sạch):")
print(fd.most_common(10))

# =====================================================================
# ## PHẦN 03: RÚT TRÍCH VÀ XỬ LÝ VĂN BẢN
# =====================================================================

# ### 04. Rút trích văn bản từ trang HTML
print("\n--- ### 07. RÚT TRÍCH VĂN BẢN TỪ TRANG HTML ---")
url = "http://news.bbc.co.uk/2/hi/health/2284783.stm"
html = request.urlopen(url).read().decode('utf8')
raw = BeautifulSoup(html, "lxml").get_text()
tokens = word_tokenize(raw)
print("50 ký tự đầu tiên trích xuất từ web:", raw[:50])

# =====================================================================
# ## PHẦN 04: PHÂN TÍCH CẢM XÚC (SENTIMENT ANALYSIS)
# =====================================================================

# ### 05. Huấn luyện Naïve Bayes để phân loại cảm xúc Movie Reviews
print("\n--- ### 08. PHÂN TÍCH CẢM XÚC (NAÏVE BAYES) ---")
reviews = nltk.corpus.movie_reviews
documents = [(list(reviews.words(fileid)), category)
             for category in reviews.categories()
             for fileid in reviews.fileids(category)]
random.seed(42)
random.shuffle(documents)

# Định nghĩa đặc trưng (features)
all_words = nltk.FreqDist(w.lower() for w in reviews.words())
word_features = list(all_words)[:2000]

def document_features(document, word_features):
    document_words = set(document)
    features = {}
    for word in word_features:
        features['{}'.format(word)] = (word in document_words)
    return features

featuresets = [(document_features(d, word_features), c) for (d,c) in documents]
train_set, test_set = featuresets[1500:], featuresets[:500]

classifier = nltk.NaiveBayesClassifier.train(train_set)
print(f"Độ chính xác của mô hình: {nltk.classify.accuracy(classifier, test_set):.3f}")
print("10 đặc trưng thông tin nhất:")
classifier.show_most_informative_features(10)

# =====================================================================
# ## PHẦN 05: BÀI TẬP ÁP DỤNG (WORDNET & NAMES)
# =====================================================================

# ### 06. So sánh độ tương đồng của từ (WordNet)
print("\n--- BÀI TẬP: ĐỘ TƯƠNG ĐỒNG GIỮA CÁC TỪ ---")
danh_tu_1, danh_tu_2 = wordnet.synset('dog.n.01'), wordnet.synset('cat.n.01')
print(f"Độ tương đồng (Dog vs Cat): {danh_tu_1.path_similarity(danh_tu_2):.2f}")

# ### 07. Xử lý tập dữ liệu Names
print("\n--- BÀI TẬP: PHÂN TÍCH TÊN (NAMES CORPUS) ---")
ten_nam = names.words('male.txt')
ten_nu = names.words('female.txt')
print(f"Số lượng Nam: {len(ten_nam)}, Số lượng Nữ: {len(ten_nu)}")

# Tạo mảng mới chứa: (Kí tự cuối của tên, Nhãn giới tính)
mang_ki_tu_cuoi = ([(name[-1].lower(), 'male') for name in ten_nam] +
                   [(name[-1].lower(), 'female') for name in ten_nu])
print("10 phần tử đầu tiên của mảng kí tự cuối (Kí tự, Nhãn):")
for item in mang_ki_tu_cuoi[:10]:
    print(item)

```
